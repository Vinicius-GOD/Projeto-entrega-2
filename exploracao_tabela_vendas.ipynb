# Exploração da tabela de vendas


# 1. Leitura da planilha


```python
import pandas as pd
from pathlib import Path

arquivo = Path("tabela_vendas_ZePequeno_corrigido.xlsx")
df = pd.read_excel(arquivo)

print("Arquivo:", arquivo.name)
df.head()
```

## 2. Colunas, tamanho e tipos

```python
print("Colunas:")
print(df.columns.tolist())
print()
print("Formato (linhas, colunas):", df.shape)
print()
print("Tipos:")
print(df.dtypes)
```

## 4. Regiões


```python
print("Regiões distintas:", df["Região"].nunique())
print()
print("Lista:")
print(sorted(df["Região"].dropna().unique()))
print()
df["Região"].value_counts(dropna=False)
```

## 5. Canais de venda

```python
print("Canais distintos:", df["Canal"].nunique())
print()
print("Lista:")
print(sorted(df["Canal"].dropna().unique()))
print()
df["Canal"].value_counts(dropna=False)
```

## 6. Produtos

```python
print("Produtos distintos:", df["Produto"].nunique())
print()
print("Lista:")
print(sorted(df["Produto"].dropna().unique()))
print()
df["Produto"].value_counts(dropna=False)
```

## 7. Datas e valores

```python
print("Data mínima:", df["Data"].min())
print("Data máxima:", df["Data"].max())
print()
print("Vendas por ano:")
print(df["Data"].dt.year.value_counts().sort_index())
```
